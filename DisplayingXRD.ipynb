```python
import pandas as pd
import xarray as xr
import numpy as np
import matplotlib.pyplot as plt
import matplotlib.ticker as ticker
from pathlib import Path
```

```python
data = pd.read_csv('./data/XRD/09-29-CMCS-jacs2.csv')
data.columns = ['#twotheta','yobs','ycal','bkg','diff']
data
```

```python
print("max:",data['yobs'].max())
```

```python
#Creating a figure based on the file uploaded earlier

fig, ax = plt.subplots(1,1,figsize=(11,5),sharex=True,sharey=True) # figure parameters and size

# labeling the axes and title
ax.set_xlabel('2Theta (degree)')
fig.supylabel('Intensity (a.u.)')
ax.set_title('XRD of the JACS 2 product')
ax.plot(data['#twotheta'],data['yobs'])
ax.set_xlim(5,45)
ax.set_ylim(0,7500)
ax.yaxis.set_minor_locator(ticker.AutoMinorLocator())
ax.xaxis.set_minor_locator(ticker.AutoMinorLocator())

figure = Path('./Figures/XRD_JACS2_product.png')
if figure.exists():
    print('figure already exists')
else:
    plt.savefig(figure)
plt.show()
```
