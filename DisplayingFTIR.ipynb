```python
import pandas as pd
import xarray as xr
import numpy as np
import matplotlib.pyplot as plt
import matplotlib.ticker as ticker
from pathlib import Path
```

```python
data = pd.read_csv('./data/FTIR/CMCSJacs2.1.dpt',sep='\t',names=['Wavelength','Absorbance'])
data
```

```python
data['Absorbance'].max()
```

```python
#Creating a figure based on the file uploaded earlier

fig, ax = plt.subplots(1,1,figsize=(11,5),sharex=True,sharey=True) #figure parameters and size

#Labeling the axes and title
ax.set_xlabel('Wavelength (cm^-1)')
fig.supylabel('Absorbance')
ax.set_title('FTIR Spectrum of JACS 2 Synthesis')
ax.plot(data['Wavelength'],data['Absorbance'])
ax.set_xlim(4000,400)
ax.set_ylim(0,0.275)
ax.yaxis.set_minor_locator(ticker.AutoMinorLocator())
ax.xaxis.set_minor_locator(ticker.AutoMinorLocator())


figure = Path('./Figures/FTIR_JACS_2_product.png')
if figure.exists():
    print('figure already exists')
else:
    plt.savefig(figure)
plt.show()
```
